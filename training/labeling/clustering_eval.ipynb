# Clustering Evaluation

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
input_path = "../../data/edgelists"
edgelist_full = pd.read_csv(f"{input_path}/edgelist_full.csv")
print(f"Edgelist full shape: {edgelist_full.shape}")

edgelist_topk = pd.read_csv(f"{input_path}/edgelist_top10.csv")
print(f"Edgelist Top 10 shape: {edgelist_topk.shape}")

edgelist_high_score = pd.read_csv(f"{input_path}/edgelist_high_scores.csv")
print(f"Edgelist full shape: {edgelist_high_score.shape}")
```

## Validate User IDs

```python
dfs = [edgelist_full, edgelist_topk, edgelist_high_score]
for df in dfs:
    print(f"Unique users: {df.user_anchor.nunique()}")
```

This is good - we want the user counts to match

## What does similarity score look like?

This is for all users - to understand where we should even put the cutoff

```python
plt.figure(figsize=(8, 5))
sns.histplot(
    edgelist_full["similarity_score"],
    kde=True,
    bins=20,
    color="skyblue",
    edgecolor="black",
)
plt.title("Distribution of Similarity Score (0 to 1)", fontsize=14)
plt.xlabel("Similarity Score", fontsize=12)
plt.ylabel("Frequency", fontsize=12)
plt.xlim(0, 1)
plt.show()
```

## Average / Min / Max similarity score for top k users

```python
plt.figure(figsize=(8, 5))
sns.histplot(
    edgelist_topk["similarity_score"],
    kde=True,
    bins=20,
    color="skyblue",
    edgecolor="black",
)
plt.loglog()
plt.title("Distribution of Similarity Score (0 to 1)", fontsize=14)
plt.xlabel("Similarity Score", fontsize=12)
plt.ylabel("Frequency", fontsize=12)
plt.xlim(0.6, 1)
plt.show()
```
